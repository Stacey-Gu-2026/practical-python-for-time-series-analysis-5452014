# Join and preprocess multiple time series data

## Load data

```python
path = '../../data/data_FRED/CORESTICKM159SFRBATL.csv'
```

```python
import pandas as pd

pdf = pd.read_csv(filepath_or_buffer=path, index_col='observation_date')
```

```python
pdf.plot(title='FRED Data - first PDF (Inflation)')
```

## Parse dates to join time series

```python
path = '../../data/data_FRED/CORESTICKM159SFRBATL.csv'
pdf1 = pd.read_csv(filepath_or_buffer=path, 
                   index_col='observation_date',
                   parse_dates=True)

pdf1
```

```python
path = '../../data/data_FRED/MORTGAGE30US.csv'
pdf2 = pd.read_csv(filepath_or_buffer=path, 
                   index_col='observation_date',
                   parse_dates=True)

pdf2
```

```python
pdf1.index
```

```python
pdf_join = pd.concat([pdf1, pdf2], axis=1, )
pdf_join.tail(50).style
```

```python
pdf_join.plot(title='FRED Data - Outer Join')
```

## Inner join vs outer join

```python
pdf_join_inner = pd.concat([pdf1, pdf2], axis=1, join='inner')
pdf_join_inner.tail(10).style
```

```python
pdf_join_inner.plot(title="FRED Data - Inner Join")
```

## Interpolate missing values

```python
pdf_join = pd.concat([pdf1, pdf2], axis=1, join='outer')
pdf_join.tail(20).style
```

```python
pdf_join_interpl = pdf_join.interpolate(method='linear')
pdf_join_interpl.tail(20).style
```

```python
pdf_join_interpl.plot(title = 'FRED Data - Linear Interpolated')
```

## Iterate for seamless loading of multiple datasets

```python
paths = [
    '../../data/data_FRED/CORESTICKM159SFRBATL.csv',
    '../../data/data_FRED/MORTGAGE30US.csv',
    '../../data/data_FRED/FEDFUNDS.csv',
    '../../data/data_FRED/T10YIE.csv',
    '../../data/data_FRED/UNRATE.csv'
]

pdfs = []
for path in paths:
    pdf_i = pd.read_csv(path, index_col=0, parse_dates=True)
    pdfs.append(pdf_i)
pdfs
```

```python
pdfs_concat = pd.concat(pdfs, axis=1, join='outer')
pdfs_concat.tail(20).style
```

```python
pdfs_interpl = pdfs_concat.interpolate(method='linear')
pdfs_interpl.tail(20).style
```

```python
pdfs_interpl.plot(title = 'FRED Data - multiple datasets, liner interplotated')
```

## Export data

```python
pdfs_interpl.index.name = 'date'
pdfs_interpl
```

```python
pdfs_interpl_renamed = pdfs_interpl.rename(columns={
    'CORESTICKM159SFRBATL': 'CPI',
    'MORTGAGE30US': 'MR',
    'FEDFUNDS': 'FFR',
    'T10YIE': 'INF10',
    'UNRATE': 'UNRT'
})
pdfs_interpl_renamed
```

```python
pdfs_interpl_renamed.plot(title = 'FRED Data - PDFS Liner Interpl renamed')
```

```python
pdfs_interpl_renamed.to_csv('../../data/data_FRED/FRED_joined.csv')
pdfs_interpl_renamed.to_parquet('../../data/data_FRED/FRED_joined.parquet')
```
